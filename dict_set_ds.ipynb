```python
# dict
1.dict is collection of items enclosed by {} -item is a key:value pair
```

```python
info={"name":"shiva","salary":10000.9,"adress":hyd","age":25}
print(info)
print(type(info))
```

```python
# read data and update data from dict
info{"name":"shiva","salary":12345.90."adress":hyd","age":25}
print(info)
print(info["adress"])
```

```python
info={"name":"shiva","salary":10000.9,"adress":hyd","age":25}
print(info)
info["adress"]=banglore"
print(info)
```

```python
# methods or functions in dict
print(dir(dict))
```

```python
info{"name":"shiva","salary":12345.90."adress":hyd","age":25}
print(info)
print(info.keys())
print(info.values())
print(info.items())
```

```python
#update()
info{"name":"shiva","salary":12345.90."adress":hyd","age":25}
print(info)
info.update({"salary":3000000.90})
print(info)
```

```python
#pop()
info{"name":"shiva","salary":12345.90."adress":hyd","age":25}
print(info)
info.pop("salary")
print(info)
```

# sets
1.set is a collection of unorder unique elements enclosed by {}

```python
# create set
a=(2,23,1,4,5,6,7,1,2,3}
print(a)
print(type(a))
```

```python
# methods or functions
print(dir(set))
```

```python
# add()
a={20,30,40,50}
print(a)
a.add(100)
print(a)
```

```python
#difference
a={1,2,3,4}
b={3,4,5,6}
c=a.difference(b)
print(c)
```

```python
#union
a={1,2,3}
b={3,4,5}
c=a.union(b)
print(c)
```

```python
#discard
a={1,2,3,4}
print(a)
a.discard(3)
print(a)
```
