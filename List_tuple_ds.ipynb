```python
# concatinate,repeating, accessing elements in tuple
t1=(10,20,30)
t2=(40,50,60)
result=t1+t2
print(result)
result=t1*3
print(result)
print(t1[0])
print(t2[-1])
```

```python
# tuple methods
print(dir(t1))
print(help(tuple))
```

```python
# tuple packing and unpacking
t=(10,20,30,'abhishek',25.5)
a,b,c,d,e=t 
print(a)
print(b)
print(c)
print(d)
print(e)
```

```python
# tuple immutability
t=(10,20,30,40,50)
print(t)
t=(10,20,35,40,50)
print(t)
```

```python
# tuple with single element
t1=(10,)
t2=(20)
print(type(t1))  
print(type(t2))  
```

```python
# tuple methods-count() and index()
t=(10,20,30,10,40,10,50)
print(t.count(10))  
print(t.index(30))  
```

```python
# nested tuple
t1=(10,20,(30,40),50)
print(t1)
print(t1[2])
print(t1[2][0])
```

```python
# convert list to tuple and tuple to list
fruits_list=['apple','banana','mango']
fruits_tuple=tuple(fruits_list)
print(fruits_tuple)
numbers_tuple=(10,20,30,40)
numbers_list=list(numbers_tuple)
print(numbers_list)
```
