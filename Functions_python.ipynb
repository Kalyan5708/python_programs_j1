````python
#write a pragram to sum of two numbers using function
def add_numbers():
    input1=int(input("enter first number:"))
    input2=int(input("enter second number:"))
    result=input1+input2
    return result
print("sum of two numbers=",add_numbers())
````

````python
#write a program to print addition of two strings
def add_strings():
    a = input("Enter first string: ")
    b = input("Enter second string: ")
    result = a + b
    return result
print("addition of two strings=", add_strings())
````

````python
#write a program to print profile data function with arguments
def profile_data(name, age, city):
    print("Name:", name)
    print("Age:", age)  
    print("City:", city)
    return
print(profile_data("Shiva", 25, "New York"))
print(profile_data("Anita", 30, "Los Angeles"))
````
